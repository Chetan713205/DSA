```python
from pydantic import BaseModel

class Person(BaseModel):
    name:str
    age:int
    city:str 

person=Person(name="Sumit Narayan", age="33", city="Bengaluru")
print(person)     ## does the automatic typecasting
```

```python
try:
    person=Person(name="Sumit Narayan", age="33", city=23)
    print(person)
except Exception as e:
    print(e)
```

```python
from typing import  Optional
class Employee(BaseModel):
    id: int
    name: str 
    dept: str 
    salary: Optional[float]=20000     # optional with default value as 20000
    is_active: Optional[bool]=True    # optional with default value as true

emp=Employee(id=1, name="Suresh", dept="Mantainance")
print(emp)
```

```python
class ClassRoom(BaseModel):
    cls_no: int 
    students: list[str]
    capacity: int 

cls1=ClassRoom(cls_no=20, students=("Sunil", "Ravi", "Umesh"), capacity=20)  # typecast done
print(cls1)
```

```python
class Address(BaseModel):
    street: str 
    city: str 
    zip_code: int

class Customer(BaseModel):
    cust_id: int 
    name: str 
    address: Address

c1= Customer(cust_id=10, 
             name="Dilip Kumar", 
             address={"street": "Pachna Road", "city": "Lakhisarai", "zip_code": 654985})

print(c1)
```

```python
from pydantic import BaseModel, Field
class Item(BaseModel):
    name: str=Field(min_length=2, max_length=50)
    price: float=Field(gt=0, le=1000)
    quantity: int=Field(ge=0)

item=Item(name="Book", price=299.50, quantity=20)
print(item)
```

```python
from pydantic import BaseModel, Field
class User(BaseModel):
    username: str=Field(..., description="Unique username for user")
    age: int=Field(default=18, description="user age, default is 18")
    email: str=Field(default_factory=lambda: "user@example.com", description="default mail")

u1= User(username="Deepal")
print(u1)
```

```python
print(User.model_json_schema())
```
